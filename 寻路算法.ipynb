```python
# astar.py
import heapq
import math

def heuristic(a, b):
    """启发函数:两点间欧几里得距离。
    注意——这里需要节点的坐标,所以图里每个节点要带 (x,y)。
    下面 positions 是额外加的坐标信息。"""
    ax, ay = positions[a]
    bx, by = positions[b]
    return math.hypot(bx - ax, by - ay)

def astar(graph, start, goal):
    g = {start: 0}
    parent = {start: None}
    # ★ 唯一改动 1:优先队列按 f = g + h 排序,而不是 g
    pq = [(heuristic(start, goal), start)]
    visited = set()
    stepCnt = 0

    while pq:
        f_u, u = heapq.heappop(pq)      # f 最小的先出
        stepCnt+=1
        
        if u in visited:
            continue
        visited.add(u)

        if u == goal:
            path = []
            node = goal
            while node is not None:
                path.append(node)
                node = parent[node]
            return g[goal], path[::-1], stepCnt

        for v, w in graph.get(u, {}).items():
            new_g = g[u] + w
            if v not in g or new_g < g[v]:
                g[v] = new_g
                parent[v] = u
                # ★ 唯一改动 2:推入堆时用 f = new_g + h,而不是 new_g
                f_v = new_g + heuristic(v, goal)
                heapq.heappush(pq, (f_v, v))

    return None, [], None


# 扩展地图:除了邻接表,还要有节点坐标(用于算启发函数)
graph = {
    "入口":  {"主道": 5},
    "主道":  {"入口": 5, "岔口A": 3, "岔口B": 6},
    "岔口A": {"主道": 3, "货架1": 4},
    "岔口B": {"主道": 6, "货架2": 2, "货架3": 7},
    "货架1": {"岔口A": 4},
    "货架2": {"岔口B": 2, "货架3": 3},
    "货架3": {"岔口B": 7, "货架2": 3},
}

# ★ 新增:每个节点的物理坐标(米)
positions = {
    "入口":  (0, 0),
    "主道":  (5, 0),
    "岔口A": (5, 3),
    "岔口B": (11, 0),
    "货架1": (5, 7),
    "货架2": (11, 2),
    "货架3": (14, 2),    # 货架3 在最右边
}

fpath = astar(graph, "入口", "货架3")
print(f"AStar: 代价={fpath[0]}, 弹出节点数={fpath[2]}")


def dijkstra(graph, start, goal):
    g = {start: 0}
    parent = {start: None}
    # ★ 唯一改动 1:优先队列按 f = g + h 排序,而不是 g
    pq = [(0, start)]
    visited = set()
    stepCnt = 0

    while pq:
        f_u, u = heapq.heappop(pq)      # f 最小的先出
        stepCnt+=1
        
        if u in visited:
            continue
        visited.add(u)

        if u == goal:
            path = []
            node = goal
            while node is not None:
                path.append(node)
                node = parent[node]
            return g[goal], path[::-1], stepCnt

        for v, w in graph.get(u, {}).items():
            new_g = g[u] + w
            if v not in g or new_g < g[v]:
                g[v] = new_g
                parent[v] = u
                f_v = new_g
                heapq.heappush(pq, (f_v, v))

    return None, [], None

fpath = dijkstra(graph, "入口", "货架3")
print(f"Dijkstra: 代价={fpath[0]}, 弹出节点数={fpath[2]}")
```
